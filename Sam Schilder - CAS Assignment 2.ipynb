# Exercise 1: Analytical solution of linear systems (3 points)
For linear systems $\dot{x} = A x$, we can often analytically compute the solution.

A) Using separation of variables, write down an analytical solution of the ordinary differential equation
$$\dfrac{dx}{dt} = \lambda x.$$
Make sure to express the solution with the initial condition.

B) Using this solution, plot the dynamics of the system over time given an initial condition and value of $\lambda$.

## Solution 1:


$$\frac{dx}{dt} = \lambda x$$

$$\frac{dx}{x} = \lambda \, dt$$

$$\int \frac{dx}{x} = \int \lambda \, dt$$

$$\ln|x| = \lambda t + c$$

$$x = e^{\lambda t} e^{c}$$

$$x = x_0 e^{\lambda t}$$

```python
f = lambda x0, l, t: x0 * jnp.exp(l*t)  # analytical solution

x0 = 50  # initial condition
l = 0.2  # lambda

t = jnp.linspace(0, 10, 100)  # time
x  = f(x0, l, t)

# plot
plt.plot(t, x)
plt.xlabel('Time (t)')
plt.ylabel('System State (x)')
plt.title(rf'Analytical Solution with $\lambda$={l}, $x_0$={x0}')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 2.2:
Create a numerical approximation of the simple harmonic oscillator (i.e., the harmonic oscillator with a damping coefficient of zero) using the Euler method: $$  x_{n+1}=x_{n}+\Delta_t f(t_{n},x_{n}) $$

Solve the system forward in the range $t \in [0, 50]$ using varying step sizes, given some (non-zero) initial condition. Also show the analytical solution.

Comment on what you see, and explain your observations.

## Solution 2.2:

```python
def simple_harmonic_oscillator(X, k):
    """
    Simple harmonic oscillator model
    X: state vector [x, xdot]
    k: omega
    """
    x, xdot = X
    return jnp.array([xdot, -k*x])
```

```python
def euler(f, x, t, dt):
    """
    Euler approximation

    f: state equation function
    t: time at which the function is evaluated
    x: state vector at time k
    dt: step size

    Returns the state at time k+1
    """

    return x + dt * f(x, t)
```

```python
# General Euler function
def get_euler(f, dt):
    def euler_scan_fn(carry, t):

        return carry + dt * f(carry, t), carry + dt * f(carry, t)

    return euler_scan_fn
```

```python
# Visualize the harmonic oscillator integrated with Euler for different step sizes
import time
color1 = 'r--'
color2_4 = ['tab:blue', 'tab:orange', 'tab:purple','tab:cyan']
# parameters
t0 = 0  # starting time
tn = 50  # end time
dts = 0.01, 0.02, 0.05, 0.3  # stepsize
t_analytical = jnp.linspace(t0, tn, 1000)
x0 = jnp.array([1.0, 0.0])

x  = f1(t); v = f2(t)
k = 1

# simulation
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(t_analytical, jnp.cos(t_analytical), color1, label=r'Analytical ($\cos(t)$)', alpha=0.7)
axes[1].plot(jnp.cos(t_analytical), -jnp.sin(t_analytical), color1, label=r'Analytical ($\cos(t)$)', alpha=0.7)

for i, dt in enumerate(dts):  # loop over all stepsizes
    num_steps = int((tn-t0)/dt)
    t = jnp.linspace(t0, tn, num_steps)  # time range

    f = lambda X, t: simple_harmonic_oscillator(X, k)
    step_fn = get_euler(f, dt)
    _, trajectory = jax.lax.scan(step_fn, x0, t)

    # plot
    axes[0].plot(t, trajectory[:,0], color2_4[i], label=rf'Euler dt = {dt}')
    axes[1].plot(trajectory[:,0], trajectory[:,1], color2_4[i], label=rf'Euler dt = {dt}')

axes[0].set_title('Time Domain (Position vs Time)')
axes[0].set_xlabel('Time (t)')
axes[0].set_ylabel('Position (x)')
axes[0].set_ylim([-3, 3])
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].set_title('State Space (Velocity vs Position)')
axes[1].set_xlabel('Position (x)')
axes[1].set_ylabel('Velocity (v)')
axes[1].set_xlim([-3, 3])
axes[1].set_ylim([-3, 3])
axes[1].set_aspect('equal', 'box')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

It can clearly be seen in the figure that Euler's method both position and velocity keep increasing over time. This is because Euler integration is an explicit method, making it unstable. What can be observed in the state-space figure is that for dt=0.3, there is a linear approximation for each step, which constantly overshoots the actual trajectory.

## 2.3

Write a RK4 solver and use it to simulate the simple harmonic oscillator. Using different step sizes, what do you observe? Compare your solution with Euler’s method. The RK4 solvers works as follows:

$$x_{n+1} = x_n + \frac{\Delta_t}{6}(k_1 + 2k_2 + 2k_3 + k_4)$$
$$t_{n+1} = t_n + \Delta_t$$
with
$$k_1 = f(t_n, x_n)$$
$$k_2 = f(t_n + \frac{\Delta_t}{2}, x_n + \Delta_t\frac{k_1}{2})$$
$$k_3 = f(t_n + \frac{\Delta_t}{2}, x_n + \Delta_t\frac{k_2}{2})$$
$$k_4 = f(t_n + \Delta_t, x_n + \Delta_t k_3)$$

## Solution 2.3

```python
# General RK4 function
def get_rk4(f, dt):
    def rk4_scan_fn(carry, t):
        k1 = f(carry, t)
        k2 = f(carry + dt*k1/2, t + dt/2)
        k3 = f(carry + dt*k2/2, t + dt/2)
        k4 = f(carry + dt*k3, t + dt)
        carry = carry + dt/6 * (k1 + 2 * k2 + 2 * k3 + k4)
        return carry, carry
    
    return rk4_scan_fn
```

```python
#Visualize the harmonic oscillator integrated with RK4 for different step sizes
# parameters
t0 = 0  # starting time
tn = 50  # end time
dts = 0.01, 0.02, 0.05, 0.3, 1  # stepsize
t_analytical = jnp.linspace(t0, tn, 1000)
x0 = jnp.array([1.0, 0.0])

x  = f1(t); v = f2(t)

# simulation
fig, axes = plt.subplots(len(dts), 2, figsize=(15,12))

for i, dt in enumerate(dts):  # loop over all stepsizes
    num_steps = int((tn-t0)/dt)
    t = jnp.linspace(t0, tn, num_steps)  # time range

    f = lambda X, t: simple_harmonic_oscillator(X, k)

    # euler's method
    step_fn_euler = get_euler(f, dt)
    _, trajectory_euler = jax.lax.scan(step_fn_euler, x0, t)

    # RK4 method
    step_fn_RK4 = get_rk4(f, dt)
    _, trajectory_RK4 = jax.lax.scan(step_fn_RK4, x0, t)

    # plot
    axes[i,0].plot(t_analytical, jnp.cos(t_analytical), label=r'Analytical ($\cos(t)$)', alpha=0.7)
    axes[i,0].plot(t, trajectory_euler[:,0], label="Euler's method")
    axes[i,0].plot(t, trajectory_RK4[:,0], 'r--', label="RK4 integrator", alpha=0.7)

    axes[i,1].plot(jnp.cos(t_analytical), -jnp.sin(t_analytical), label=r'Analytical ($\cos(t)$)', alpha=0.7)
    axes[i,1].plot(trajectory_euler[:,0], trajectory_euler[:,1], label="Euler's method")
    axes[i,1].plot(trajectory_RK4[:300,0], trajectory_RK4[:300,1], 'r--', label="RK4 integrator", alpha=0.7)

    axes[i,0].set_title(f'dt = {dt}')
    axes[i,0].set_xlabel('Time (t)')
    axes[i,0].set_ylabel('Position (x)')
    axes[i,0].set_ylim([-3, 3])
    axes[i,0].legend()
    axes[i,1].legend(loc='upper left')
    axes[i,0].grid(True, alpha=0.3)

    axes[i,1].set_title(f'State Space (dt = {dt})')
    axes[i,1].set_xlabel('Position (x)')
    axes[i,1].set_ylabel('Velocity (v)')
    axes[i,1].set_xlim([-3, 3])
    axes[i,1].set_ylim([-3, 3])
    axes[i,1].set_aspect('equal', 'box')
    axes[i,1].legend(loc='upper right')
    axes[i,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

When comparing the RK4 method to the Euler method, we can observe that it clearly works better for regular sized time steps (e.g. dt = 0.05). However, also RK4 becomes inaccurate once the time steps become larger. We can see a clear phase shift for dt=1, with an inwards spriral trajectory for the state space, indicating that position and velocity may decrease over time. The reason RK4 is more accurate than the Euler method is because it is an 4th order method, making it more sensitive to the curved state space trajectory.
